```python
import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem
import matplotlib.pyplot as plt
import seaborn as sns

# ===== KROK 1: ZAŁADUJ DANE =====
print("=" * 80)
print("PRZYGOTOWANIE SMILES - MORGAN FINGERPRINTY")
print("=" * 80)

csv_path = r"C:\Users\slast\PYTHON\0_projekty do portfolio\07_pytorch_cl\data\procced\data_preprocessed.csv"
df = pd.read_csv(csv_path)

print(f"\nDane załadowane:")
print(f"  Shape: {df.shape}")
print(f"  Kolumny: {df.columns.tolist()}")
print(f"\nPierwsze 5 SMILES:")
for i in range(5):
    print(f"  {i}: {df['SMILES'].iloc[i]}")

# ===== KROK 2: PARAMETRY MORGAN FINGERPRINTS =====
print(f"\n{'-' * 80}")
print(f"PARAMETRY MORGAN FINGERPRINTS")
print(f"{'-' * 80}")

# Parametry do testowania
RADIUS = 2  # Rozmiar neighborhood
N_BITS = 2048  # Długość fingerprinta (power of 2)

print(f"\nRadius: {RADIUS}")
print(f"  (określa rozmiar otoczenia atomu)")
print(f"N_BITS: {N_BITS}")
print(f"  (długość binarnego wektora)")
print(f"\nOutput: wektor {N_BITS} zer i jedynek na każdą cząsteczkę")

# ===== KROK 3: GENEROWANIE FINGERPRINTS =====
print(f"\n{'-' * 80}")
print(f"GENEROWANIE MORGAN FINGERPRINTS")
print(f"{'-' * 80}")

def smiles_to_morgan_fp(smiles, radius=RADIUS, n_bits=N_BITS):
    """
    Konwertuje SMILES na Morgan fingerprint (bitvector)
    
    Args:
        smiles: string SMILES
        radius: rozmiar neighborhood (default: 2)
        n_bits: długość fingerprinta (default: 2048)
    
    Returns:
        numpy array o rozmiarze (n_bits,) lub None jeśli błąd
    """
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            return None
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius, nBits=n_bits)
        return np.array(fp, dtype=np.float32)
    except Exception as e:
        print(f"  ⚠️  Błąd przy SMILES '{smiles}': {str(e)}")
        return None

print(f"\nGenerowanie fingerprints dla {len(df)} cząsteczek...")
print(f"(to może chwilę potrwać...)\n")

fingerprints = []
failed_indices = []

for idx, smiles in enumerate(df['SMILES']):
    if idx % 2000 == 0:
        print(f"  Przetworzono: {idx}/{len(df)}")
    
    fp = smiles_to_morgan_fp(smiles)
    
    if fp is not None:
        fingerprints.append(fp)
    else:
        fingerprints.append(None)
        failed_indices.append(idx)

# ===== KROK 4: WERYFIKACJA =====
print(f"\n{'-' * 80}")
print(f"WERYFIKACJA")
print(f"{'-' * 80}")

n_success = sum(1 for fp in fingerprints if fp is not None)
n_failed = len(failed_indices)

print(f"\nSuccessfull fingerprints: {n_success}")
print(f"Failed fingerprints: {n_failed}")
print(f"Success rate: {(n_success / len(df)) * 100:.2f}%")

if n_failed > 0:
    print(f"\nIndeksy SMILES które się nie konwertowały:")
    print(f"  {failed_indices[:20]}")  # Pokaż pierwszych 20
    
    if n_failed <= 10:
        print(f"\nProblematyczne SMILES:")
        for idx in failed_indices:
            print(f"  {idx}: {df['SMILES'].iloc[idx]}")

# ===== KROK 5: USUŃ FAILED SMILES =====
print(f"\n{'-' * 80}")
print(f"CZYSZCZENIE DANYCH")
print(f"{'-' * 80}")

if n_failed > 0:
    print(f"\nUsuwacie wiersze gdzie SMILES nie mogły być przetworzony...")
    df_clean = df.drop(failed_indices).reset_index(drop=True)
    fingerprints = [fp for fp in fingerprints if fp is not None]
    
    print(f"\nBefore: {len(df)} wierszy")
    print(f"After: {len(df_clean)} wierszy")
    print(f"Usunięto: {n_failed} wierszy")
else:
    df_clean = df.copy()

# ===== KROK 6: KONWERSJA NA ARRAY =====
print(f"\n{'-' * 80}")
print(f"KONWERSJA NA ARRAY")
print(f"{'-' * 80}")

X_morgan = np.array(fingerprints, dtype=np.float32)

print(f"\nMorgan fingerprints shape: {X_morgan.shape}")
print(f"  Liczba próbek: {X_morgan.shape[0]}")
print(f"  Długość fingerprinta: {X_morgan.shape[1]}")

print(f"\nDatype: {X_morgan.dtype}")
print(f"Memory: {X_morgan.nbytes / 1024**2:.2f} MB")

# ===== KROK 7: STATYSTYKA FINGERPRINTS =====
print(f"\n{'-' * 80}")
print(f"STATYSTYKA FINGERPRINTS")
print(f"{'-' * 80}")

print(f"\nLiczba bitów na 1 (średnia na próbkę):")
bits_per_sample = X_morgan.sum(axis=1)
print(f"  Mean: {bits_per_sample.mean():.2f}")
print(f"  Std: {bits_per_sample.std():.2f}")
print(f"  Min: {bits_per_sample.min():.0f}")
print(f"  Max: {bits_per_sample.max():.0f}")

print(f"\nOdsetek aktywnych bitów (ogółem):")
active_bits = X_morgan.sum() / X_morgan.size * 100
print(f"  {active_bits:.2f}%")

# ===== KROK 8: WIZUALIZACJA =====
print(f"\n{'-' * 80}")
print(f"WIZUALIZACJA")
print(f"{'-' * 80}")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Morgan Fingerprints - Analiza', fontsize=14, fontweight='bold')

# Histogram - liczba bitów na próbkę
axes[0, 0].hist(bits_per_sample, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
axes[0, 0].axvline(bits_per_sample.mean(), color='red', linestyle='--', linewidth=2, 
                   label=f'Mean: {bits_per_sample.mean():.1f}')
axes[0, 0].set_xlabel('Liczba aktywnych bitów', fontsize=11)
axes[0, 0].set_ylabel('Frequency', fontsize=11)
axes[0, 0].set_title('Rozkład aktywnych bitów na próbkę', fontsize=12, fontweight='bold')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Heatmap - pierwsze 50 samples x 100 bitów
axes[0, 1].imshow(X_morgan[:50, :100], aspect='auto', cmap='binary', interpolation='nearest')
axes[0, 1].set_xlabel('Bity (0-100)', fontsize=11)
axes[0, 1].set_ylabel('Próbki (0-50)', fontsize=11)
axes[0, 1].set_title('Heatmap - Morgan FP (pierwsze 50 próbek)', fontsize=12, fontweight='bold')

# Box plot - rozkład bitów
axes[1, 0].boxplot(bits_per_sample, vert=True)
axes[1, 0].set_ylabel('Liczba aktywnych bitów', fontsize=11)
axes[1, 0].set_title('Box Plot - Aktywne bity', fontsize=12, fontweight='bold')
axes[1, 0].grid(alpha=0.3, axis='y')

# Histogram bitów (który bit jest częściej aktywny)
bit_activity = X_morgan.sum(axis=0)
axes[1, 1].hist(bit_activity, bins=50, color='lightcoral', edgecolor='black', alpha=0.7)
axes[1, 1].set_xlabel('Liczba próbek gdzie bit=1', fontsize=11)
axes[1, 1].set_ylabel('Frequency', fontsize=11)
axes[1, 1].set_title('Rozkład aktywności bitów', fontsize=12, fontweight='bold')
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
viz_path = r"C:\Users\slast\PYTHON\0_projekty do portfolio\07_pytorch_cl\visualisation\05_morgan_fingerprints_analysis.png"
plt.savefig(viz_path, dpi=300, bbox_inches='tight')
print(f"\n✅ Wykres zapisany: {viz_path}")
plt.show()

# ===== KROK 9: PRZYGOTOWANIE DO DALSZYCH KROKÓW =====
print(f"\n{'-' * 80}")
print(f"PODSUMOWANIE")
print(f"{'-' * 80}")

print(f"""
✅ Morgan Fingerprinty przygotowane:
   - Liczba próbek: {X_morgan.shape[0]}
   - Rozmiar fingerprinta: {X_morgan.shape[1]}
   - Success rate: {(n_success / len(df)) * 100:.2f}%

📊 Statystyka:
   - Aktywne bity na próbkę: {bits_per_sample.mean():.1f} ± {bits_per_sample.std():.1f}
   - Odsetek aktywnych bitów: {active_bits:.2f}%

📁 Następne kroki:
   1. Połącz Morgan FP z numerycznymi features
   2. Utwórz DataLoader
   3. Zbuduj model PyTorch
   4. Trenuj!
""")

# Zapisz zmienne do użytku w dalszych komórkach
print(f"✅ ZMIENNE DOSTĘPNE:")
print(f"   - df_clean: DataFrame z czystymi danymi")
print(f"   - X_morgan: numpy array Morgan fingerprints ({X_morgan.shape})")
print(f"   - RADIUS: {RADIUS}")
print(f"   - N_BITS: {N_BITS}")
```

```python
import pandas as pd
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader, random_split
import matplotlib.pyplot as plt

# ===== KROK 1: PRZYGOTOWANIE DANYCH =====
print("=" * 80)
print("KOMBINACJA FEATURES + MORGAN FINGERPRINTS + DATALOADER")
print("=" * 80)

# Kolumny numerycznych features
feature_cols = ['Molecular_Weight', 'Polar_Area', 'Complexity', 
                'H-Bond_Donor_Count', 'H-Bond_Acceptor_Count', 'Rotatable_Bond_Count']
target_col = 'XLogP'

# Ekstrakcja features z df_clean
X_numeric = df_clean[feature_cols].values.astype(np.float32)
y = df_clean[target_col].values.astype(np.float32)

print(f"\nNumeric features shape: {X_numeric.shape}")
print(f"Morgan fingerprints shape: {X_morgan.shape}")
print(f"Target (XLogP) shape: {y.shape}")

# ===== KROK 2: KOMBINACJA FEATURES =====
print(f"\n{'-' * 80}")
print(f"KOMBINACJA FEATURES")
print(f"{'-' * 80}")

# Połącz Morgan FP (2048) + numeric features (6)
X_combined = np.hstack([X_morgan, X_numeric])

print(f"\nGórne bity: Morgan fingerprints (2048)")
print(f"Dolne bity: Numeric features (6)")
print(f"  - Molecular_Weight (skalowany)")
print(f"  - Polar_Area (skalowany)")
print(f"  - Complexity (skalowany)")
print(f"  - H-Bond_Donor_Count (skalowany)")
print(f"  - H-Bond_Acceptor_Count (skalowany)")
print(f"  - Rotatable_Bond_Count (skalowany)")

print(f"\nFinal combined features shape: {X_combined.shape}")
print(f"  Liczba próbek: {X_combined.shape[0]}")
print(f"  Liczba features: {X_combined.shape[1]} (2048 + 6)")

# ===== KROK 3: CUSTOM DATASET CLASS =====
print(f"\n{'-' * 80}")
print(f"CUSTOM PYTORCH DATASET")
print(f"{'-' * 80}")

class MoleculeDataset(Dataset):
    """
    Custom Dataset dla danych molekularnych
    
    Features: 2054 (2048 Morgan FP + 6 numeric features)
    Target: XLogP (lipofilowość)
    """
    def __init__(self, X, y):
        """
        Args:
            X: numpy array (n_samples, 2054) - combined features
            y: numpy array (n_samples,) - target values (XLogP)
        """
        self.X = torch.FloatTensor(X)
        self.y = torch.FloatTensor(y)
        
    def __len__(self):
        return len(self.X)
    
    def __getitem__(self, idx):
        """
        Returns:
            features: tensor (2054,)
            target: scalar tensor
        """
        return self.X[idx], self.y[idx]

print(f"\n✅ MoleculeDataset class created")
print(f"   - Input: 2054 features")
print(f"   - Output: XLogP (scalar)")

# ===== KROK 4: TRAIN/VAL/TEST SPLIT =====
print(f"\n{'-' * 80}")
print(f"TRAIN/VAL/TEST SPLIT")
print(f"{'-' * 80}")

# Tworzymy dataset
dataset = MoleculeDataset(X_combined, y)

# Proporcje
train_size = 0.7  # 70%
val_size = 0.15   # 15%
test_size = 0.15  # 15%

n_train = int(len(dataset) * train_size)
n_val = int(len(dataset) * val_size)
n_test = len(dataset) - n_train - n_val

print(f"\nProporcje:")
print(f"  Train: {train_size*100:.0f}% ({n_train} próbek)")
print(f"  Val:   {val_size*100:.0f}% ({n_val} próbek)")
print(f"  Test:  {test_size*100:.0f}% ({n_test} próbek)")

# Split
train_dataset, val_dataset, test_dataset = random_split(
    dataset, 
    [n_train, n_val, n_test],
    generator=torch.Generator().manual_seed(42)  # Dla reproducibility
)

print(f"\n✅ Datasets created:")
print(f"   - Train: {len(train_dataset)}")
print(f"   - Val: {len(val_dataset)}")
print(f"   - Test: {len(test_dataset)}")
print(f"   - Total: {len(dataset)}")

# ===== KROK 5: DATALOADER =====
print(f"\n{'-' * 80}")
print(f"PYTORCH DATALOADER")
print(f"{'-' * 80}")

# Parametry
BATCH_SIZE = 32
NUM_WORKERS = 0  # 0 na Windows, więcej na Linux

# Tworzenie DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

print(f"\nBatch size: {BATCH_SIZE}")
print(f"\n✅ DataLoaders created:")
print(f"   - Train batches: {len(train_loader)}")
print(f"   - Val batches: {len(val_loader)}")
print(f"   - Test batches: {len(test_loader)}")

# ===== KROK 6: TESTOWANIE DATALOADER =====
print(f"\n{'-' * 80}")
print(f"TEST DATALOADER")
print(f"{'-' * 80}")

# Weź jeden batch z train loadera
batch_features, batch_targets = next(iter(train_loader))

print(f"\nPierwszy batch z train_loader:")
print(f"  Features shape: {batch_features.shape}")
print(f"    - Batch size: {batch_features.shape[0]}")
print(f"    - Features: {batch_features.shape[1]} (2048 Morgan + 6 numeric)")
print(f"  Targets shape: {batch_targets.shape}")

print(f"\nStatystyka batch'a:")
print(f"  Features - Mean: {batch_features.mean():.4f}, Std: {batch_features.std():.4f}")
print(f"  Targets - Mean: {batch_targets.mean():.4f}, Std: {batch_targets.std():.4f}")
print(f"  Targets - Min: {batch_targets.min():.4f}, Max: {batch_targets.max():.4f}")

# ===== KROK 7: WIZUALIZACJA TARGETU =====
print(f"\n{'-' * 80}")
print(f"WIZUALIZACJA ROZKŁADU TARGET W ZBIORACH")
print(f"{'-' * 80}")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
fig.suptitle('Rozkład XLogP w Train/Val/Test setach', fontsize=14, fontweight='bold')

# Train
y_train = np.array([y for _, y in train_dataset])
axes[0].hist(y_train, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
axes[0].axvline(y_train.mean(), color='red', linestyle='--', linewidth=2, 
                label=f'Mean: {y_train.mean():.2f}')
axes[0].set_xlabel('XLogP', fontsize=11)
axes[0].set_ylabel('Frequency', fontsize=11)
axes[0].set_title(f'TRAIN (n={len(y_train)})', fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(alpha=0.3)

# Val
y_val = np.array([y for _, y in val_dataset])
axes[1].hist(y_val, bins=50, color='lightcoral', edgecolor='black', alpha=0.7)
axes[1].axvline(y_val.mean(), color='red', linestyle='--', linewidth=2, 
                label=f'Mean: {y_val.mean():.2f}')
axes[1].set_xlabel('XLogP', fontsize=11)
axes[1].set_ylabel('Frequency', fontsize=11)
axes[1].set_title(f'VAL (n={len(y_val)})', fontsize=12, fontweight='bold')
axes[1].legend()
axes[1].grid(alpha=0.3)

# Test
y_test = np.array([y for _, y in test_dataset])
axes[2].hist(y_test, bins=50, color='lightgreen', edgecolor='black', alpha=0.7)
axes[2].axvline(y_test.mean(), color='red', linestyle='--', linewidth=2, 
                label=f'Mean: {y_test.mean():.2f}')
axes[2].set_xlabel('XLogP', fontsize=11)
axes[2].set_ylabel('Frequency', fontsize=11)
axes[2].set_title(f'TEST (n={len(y_test)})', fontsize=12, fontweight='bold')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.tight_layout()
viz_path = r"C:\Users\slast\PYTHON\0_projekty do portfolio\07_pytorch_cl\visualisation\06_data_split_distribution.png"
plt.savefig(viz_path, dpi=300, bbox_inches='tight')
print(f"\n✅ Wykres zapisany: {viz_path}")
plt.show()

# ===== KROK 8: PODSUMOWANIE =====
print(f"\n{'-' * 80}")
print(f"PODSUMOWANIE")
print(f"{'-' * 80}")

print(f"""
✅ DATALOADER GOTOWY:

📊 Dane:
   - Liczba próbek: {len(dataset)}
   - Features: 2054 (2048 Morgan FP + 6 numeric)
   - Target: XLogP

📈 Split:
   - Train: {len(train_dataset)} ({len(train_dataset)/len(dataset)*100:.1f}%)
   - Val: {len(val_dataset)} ({len(val_dataset)/len(dataset)*100:.1f}%)
   - Test: {len(test_dataset)} ({len(test_dataset)/len(dataset)*100:.1f}%)

🔄 DataLoaders:
   - Batch size: {BATCH_SIZE}
   - Train batches: {len(train_loader)}
   - Val batches: {len(val_loader)}
   - Test batches: {len(test_loader)}

📁 ZMIENNE DOSTĘPNE:
   - train_loader, val_loader, test_loader
   - train_dataset, val_dataset, test_dataset
   - dataset (pełny)
   - X_combined, y
   - BATCH_SIZE

🚀 NASTĘPNY KROK: Budowa modelu PyTorch!
""")

# Zapisz do użytku w dalszych komórkach
INPUT_SIZE = X_combined.shape[1]
print(f"✅ INPUT_SIZE dla modelu: {INPUT_SIZE}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim.lr_scheduler import ReduceLROnPlateau
import numpy as np
import matplotlib.pyplot as plt
import os
from datetime import datetime

# ===== KROK 1: DEVICE =====
print("=" * 80)
print("BUDOWA MODELU PYTORCH - TRENOWANIE")
print("=" * 80)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"\nDevice: {device}")
if device.type == 'cuda':
    print(f"GPU: {torch.cuda.get_device_name(0)}")

# ===== KROK 2: ARCHITEKTURA MODELU =====
print(f"\n{'-' * 80}")
print(f"ARCHITEKTURA MODELU")
print(f"{'-' * 80}")

class MoleculeLogPPredictor(nn.Module):
    """
    Fully Connected Neural Network do predykcji logP
    
    Architecture:
    Input (2054) → Hidden1 (512) → Hidden2 (256) → Hidden3 (128) → Output (1)
    
    Activation: ReLU
    Regularization: Dropout (0.3)
    Batch Norm: Po każdej warstwie
    """
    def __init__(self, input_size=2054, dropout_rate=0.3):
        super(MoleculeLogPPredictor, self).__init__()
        
        self.fc1 = nn.Linear(input_size, 512)
        self.bn1 = nn.BatchNorm1d(512)
        self.dropout1 = nn.Dropout(dropout_rate)
        
        self.fc2 = nn.Linear(512, 256)
        self.bn2 = nn.BatchNorm1d(256)
        self.dropout2 = nn.Dropout(dropout_rate)
        
        self.fc3 = nn.Linear(256, 128)
        self.bn3 = nn.BatchNorm1d(128)
        self.dropout3 = nn.Dropout(dropout_rate)
        
        self.fc4 = nn.Linear(128, 1)
        
        self.relu = nn.ReLU()
        
    def forward(self, x):
        """
        Forward pass
        
        Args:
            x: tensor (batch_size, 2054)
        
        Returns:
            logp: tensor (batch_size, 1)
        """
        x = self.fc1(x)
        x = self.bn1(x)
        x = self.relu(x)
        x = self.dropout1(x)
        
        x = self.fc2(x)
        x = self.bn2(x)
        x = self.relu(x)
        x = self.dropout2(x)
        
        x = self.fc3(x)
        x = self.bn3(x)
        x = self.relu(x)
        x = self.dropout3(x)
        
        x = self.fc4(x)
        
        return x

# Utwórz model
model = MoleculeLogPPredictor(input_size=INPUT_SIZE, dropout_rate=0.3)
model = model.to(device)

print(f"\nModel architecture:")
print(model)

# Liczba parametrów
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"\nParametry:")
print(f"  Total: {total_params:,}")
print(f"  Trainable: {trainable_params:,}")

# ===== KROK 3: LOSS FUNCTION I OPTIMIZER =====
print(f"\n{'-' * 80}")
print(f"LOSS FUNCTION I OPTIMIZER")
print(f"{'-' * 80}")

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-5)
scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)

print(f"\nLoss Function: MSE (Mean Squared Error)")
print(f"Optimizer: Adam")
print(f"  - Learning Rate: 0.001")
print(f"  - Weight Decay: 1e-5 (L2 regularization)")
print(f"Scheduler: ReduceLROnPlateau")
print(f"  - Factor: 0.5 (zmniejsz LR o 50%)")
print(f"  - Patience: 5 (czekaj 5 epochs bez polepszenia)")

# ===== KROK 4: FUNKCJE TRENINGOWE =====
print(f"\n{'-' * 80}")
print(f"FUNKCJE TRENINGOWE")
print(f"{'-' * 80}")

def train_epoch(model, train_loader, criterion, optimizer, device):
    """
    Trenuj model na 1 epoch
    """
    model.train()
    total_loss = 0.0
    
    for batch_features, batch_targets in train_loader:
        batch_features = batch_features.to(device)
        batch_targets = batch_targets.to(device).unsqueeze(1)
        
        # Forward pass
        predictions = model(batch_features)
        loss = criterion(predictions, batch_targets)
        
        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        total_loss += loss.item()
    
    return total_loss / len(train_loader)

def validate(model, val_loader, criterion, device):
    """
    Waliduj model na validation set
    """
    model.eval()
    total_loss = 0.0
    all_predictions = []
    all_targets = []
    
    with torch.no_grad():
        for batch_features, batch_targets in val_loader:
            batch_features = batch_features.to(device)
            batch_targets = batch_targets.to(device).unsqueeze(1)
            
            predictions = model(batch_features)
            loss = criterion(predictions, batch_targets)
            
            total_loss += loss.item()
            all_predictions.append(predictions.cpu())
            all_targets.append(batch_targets.cpu())
    
    return total_loss / len(val_loader), all_predictions, all_targets

def calculate_metrics(predictions, targets):
    """
    Oblicz metryki regresji: MAE, RMSE, R²
    """
    predictions = torch.cat(predictions).numpy().flatten()
    targets = torch.cat(targets).numpy().flatten()
    
    mae = np.mean(np.abs(predictions - targets))
    mse = np.mean((predictions - targets) ** 2)
    rmse = np.sqrt(mse)
    
    # R² = 1 - (SS_res / SS_tot)
    ss_res = np.sum((targets - predictions) ** 2)
    ss_tot = np.sum((targets - np.mean(targets)) ** 2)
    r2 = 1 - (ss_res / ss_tot)
    
    return {'mae': mae, 'mse': mse, 'rmse': rmse, 'r2': r2}

print(f"\n✅ Funkcje treningowe zdefiniowane")

# ===== KROK 5: PARAMETRY TRENOWANIA =====
print(f"\n{'-' * 80}")
print(f"PARAMETRY TRENOWANIA")
print(f"{'-' * 80}")

NUM_EPOCHS = 100
EARLY_STOPPING_PATIENCE = 15
CHECKPOINT_DIR = r"C:\Users\slast\PYTHON\0_projekty do portfolio\07_pytorch_cl\checkpoints"

# Utwórz folder na checkpointy
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

print(f"\nNum Epochs: {NUM_EPOCHS}")
print(f"Early Stopping Patience: {EARLY_STOPPING_PATIENCE}")
print(f"Checkpoint Directory: {CHECKPOINT_DIR}")

# ===== KROK 6: TRENOWANIE =====
print(f"\n{'-' * 80}")
print(f"TRENOWANIE MODELU")
print(f"{'-' * 80}")

train_losses = []
val_losses = []
best_val_loss = float('inf')
patience_counter = 0
start_time = datetime.now()

print(f"\nStart: {start_time.strftime('%Y-%m-%d %H:%M:%S')}\n")

for epoch in range(NUM_EPOCHS):
    # Train
    train_loss = train_epoch(model, train_loader, criterion, optimizer, device)
    train_losses.append(train_loss)
    
    # Validate
    val_loss, val_predictions, val_targets = validate(model, val_loader, criterion, device)
    val_losses.append(val_loss)
    
    # Scheduler
    scheduler.step(val_loss)
    
    # Early stopping
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        patience_counter = 0
        
        # Zapisz best model
        checkpoint_path = os.path.join(CHECKPOINT_DIR, 'best_model.pt')
        torch.save(model.state_dict(), checkpoint_path)
    else:
        patience_counter += 1
    
    # Печать co 5 epochs
    if (epoch + 1) % 5 == 0:
        metrics = calculate_metrics(val_predictions, val_targets)
        print(f"Epoch {epoch+1:3d}/{NUM_EPOCHS} | "
              f"Train Loss: {train_loss:.6f} | "
              f"Val Loss: {val_loss:.6f} | "
              f"Val MAE: {metrics['mae']:.4f} | "
              f"Val RMSE: {metrics['rmse']:.4f} | "
              f"Val R²: {metrics['r2']:.4f}")
    
    # Early stopping
    if patience_counter >= EARLY_STOPPING_PATIENCE:
        print(f"\n⚠️  Early stopping at epoch {epoch+1}")
        break

end_time = datetime.now()
training_time = end_time - start_time

print(f"\n✅ Trenowanie zakończone!")
print(f"Duration: {training_time}")
print(f"Best Val Loss: {best_val_loss:.6f}")

# ===== KROK 7: ZAŁADUJ BEST MODEL =====
print(f"\n{'-' * 80}")
print(f"ZAŁADUJ BEST MODEL")
print(f"{'-' * 80}")

model.load_state_dict(torch.load(os.path.join(CHECKPOINT_DIR, 'best_model.pt')))
model.eval()
print(f"\n✅ Best model załadowany")

# ===== KROK 8: FINALNA EWALUACJA =====
print(f"\n{'-' * 80}")
print(f"FINALNA EWALUACJA")
print(f"{'-' * 80}")

# Train
_, train_pred, train_targets = validate(model, train_loader, criterion, device)
train_metrics = calculate_metrics(train_pred, train_targets)

# Val
_, val_pred, val_targets = validate(model, val_loader, criterion, device)
val_metrics = calculate_metrics(val_pred, val_targets)

# Test
_, test_pred, test_targets = validate(model, test_loader, criterion, device)
test_metrics = calculate_metrics(test_pred, test_targets)

print(f"\nTRAIN:")
print(f"  MAE: {train_metrics['mae']:.4f}")
print(f"  RMSE: {train_metrics['rmse']:.4f}")
print(f"  R²: {train_metrics['r2']:.4f}")

print(f"\nVAL:")
print(f"  MAE: {val_metrics['mae']:.4f}")
print(f"  RMSE: {val_metrics['rmse']:.4f}")
print(f"  R²: {val_metrics['r2']:.4f}")

print(f"\nTEST:")
print(f"  MAE: {test_metrics['mae']:.4f}")
print(f"  RMSE: {test_metrics['rmse']:.4f}")
print(f"  R²: {test_metrics['r2']:.4f}")

# ===== KROK 9: WIZUALIZACJA =====
print(f"\n{'-' * 80}")
print(f"WIZUALIZACJA WYNIKÓW")
print(f"{'-' * 80}")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Trenowanie Modelu - Wyniki', fontsize=14, fontweight='bold')

# Loss curves
axes[0, 0].plot(train_losses, label='Train Loss', linewidth=2)
axes[0, 0].plot(val_losses, label='Val Loss', linewidth=2)
axes[0, 0].set_xlabel('Epoch', fontsize=11)
axes[0, 0].set_ylabel('Loss (MSE)', fontsize=11)
axes[0, 0].set_title('Loss Curves', fontsize=12, fontweight='bold')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Predictions vs Targets - Train
train_pred_flat = torch.cat(train_pred).numpy().flatten()
train_targets_flat = torch.cat(train_targets).numpy().flatten()
axes[0, 1].scatter(train_targets_flat, train_pred_flat, alpha=0.5, s=10)
axes[0, 1].plot([-10, 10], [-10, 10], 'r--', linewidth=2, label='Perfect Prediction')
axes[0, 1].set_xlabel('True XLogP', fontsize=11)
axes[0, 1].set_ylabel('Predicted XLogP', fontsize=11)
axes[0, 1].set_title(f'Train Predictions (R²={train_metrics["r2"]:.4f})', fontsize=12, fontweight='bold')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)
axes[0, 1].set_xlim(-10, 10)
axes[0, 1].set_ylim(-10, 10)

# Predictions vs Targets - Val
val_pred_flat = torch.cat(val_pred).numpy().flatten()
val_targets_flat = torch.cat(val_targets).numpy().flatten()
axes[1, 0].scatter(val_targets_flat, val_pred_flat, alpha=0.5, s=10, color='orange')
axes[1, 0].plot([-10, 10], [-10, 10], 'r--', linewidth=2, label='Perfect Prediction')
axes[1, 0].set_xlabel('True XLogP', fontsize=11)
axes[1, 0].set_ylabel('Predicted XLogP', fontsize=11)
axes[1, 0].set_title(f'Val Predictions (R²={val_metrics["r2"]:.4f})', fontsize=12, fontweight='bold')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)
axes[1, 0].set_xlim(-10, 10)
axes[1, 0].set_ylim(-10, 10)

# Predictions vs Targets - Test
test_pred_flat = torch.cat(test_pred).numpy().flatten()
test_targets_flat = torch.cat(test_targets).numpy().flatten()
axes[1, 1].scatter(test_targets_flat, test_pred_flat, alpha=0.5, s=10, color='green')
axes[1, 1].plot([-10, 10], [-10, 10], 'r--', linewidth=2, label='Perfect Prediction')
axes[1, 1].set_xlabel('True XLogP', fontsize=11)
axes[1, 1].set_ylabel('Predicted XLogP', fontsize=11)
axes[1, 1].set_title(f'Test Predictions (R²={test_metrics["r2"]:.4f})', fontsize=12, fontweight='bold')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)
axes[1, 1].set_xlim(-10, 10)
axes[1, 1].set_ylim(-10, 10)

plt.tight_layout()
viz_path = r"C:\Users\slast\PYTHON\0_projekty do portfolio\07_pytorch_cl\visualisation\07_training_results.png"
plt.savefig(viz_path, dpi=300, bbox_inches='tight')
print(f"\n✅ Wykres zapisany: {viz_path}")
plt.show()

# ===== KROK 10: PODSUMOWANIE =====
print(f"\n{'-' * 80}")
print(f"PODSUMOWANIE TRENOWANIA")
print(f"{'-' * 80}")

print(f"""
✅ MODEL WYTRENOWANY!

📊 Architektura:
   Input (2054) → 512 → 256 → 128 → Output (1)
   Activation: ReLU
   Regularization: Dropout (0.3) + BatchNorm
   Parameters: {trainable_params:,}

📈 Wyniki:
   Train R²: {train_metrics['r2']:.4f}, RMSE: {train_metrics['rmse']:.4f}
   Val R²:   {val_metrics['r2']:.4f}, RMSE: {val_metrics['rmse']:.4f}
   Test R²:  {test_metrics['r2']:.4f}, RMSE: {test_metrics['rmse']:.4f}

⏱️ Trenowanie:
   Epochs: {epoch + 1}/{NUM_EPOCHS}
   Duration: {training_time}

📁 Model zapisany: {os.path.join(CHECKPOINT_DIR, 'best_model.pt')}

🚀 NASTĘPNY KROK: Streamlit aplikacja!
""")

print(f"✅ ZMIENNE DOSTĘPNE:")
print(f"   - model (wytrenowany)")
print(f"   - device")
print(f"   - criterion")
print(f"   - train/val/test_metrics")
```
